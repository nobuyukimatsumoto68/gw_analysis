# GW peak signal, $N_s=24,32$ combined (single panel, $\log_{10}$ axes)

Same style as the single-volume plots (`fT_res_Ns{24,32}_claude`): the `fT_res_v2` Cell-12 layout on
linear $\log_{10}$ axes, exact arrow annotations, faint grid, tab10 palette (color $= M_B\sqrt{t_0}$,
marker $= M_B[{\rm GeV}]$). Here the 16 points are the **24c$\oplus$32c combination**.

The combination is done in $y=\ln x$ (natural, multiplicative): per volume the jackknife error is
$\sigma_y=\sigma_x/x$; then $y_{\rm comb}$ is the inverse-variance mean, $\sigma_{\rm jk}=(\sum1/\sigma_y^2)^{-1/2}$,
the systematic is $\sigma_{\rm syst}=|\ln(x_{24}/x_{32})|$, and $\sigma_{\rm tot}=\sqrt{\sigma_{\rm jk}^2+\sigma_{\rm syst}^2}$.
On these $\log_{10}$ axes the point is $\log_{10}(e^{y_{\rm comb}})$ with a **symmetric** error bar of
half-width $\sigma_{\rm tot}/\ln 10$ (the multiplicative band drawn directly in $\log_{10}$).

Axes: $x=\log_{10}(f_{\rm peak}\,[{\rm Hz}])$, $y=\log_{10}(h^2\Omega_{\rm peak})$. Data:
`gw_fin_24c_claude.log`, `gw_fin_ref_claude.dat` ($v_w=v_J$); row $=$ iMBGeV$\times4+$iMB.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# match fT_res_v2.ipynb styling (its Cell 7)
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "Helvetica",
    "font.size": 26
})
```

```python
# power-law-integrated sensitivity curves (log10 f, log10 h^2 Omega in cols 0,1)
d = "./power-law-integrated_sensitivities/"
LISA = np.loadtxt(d + "plis_LISA.dat")
BBO = np.loadtxt(d + "plis_BBO.dat")
DECIGO = np.loadtxt(d + "plis_DECIGO.dat")
ET = np.loadtxt(d + "plis_ET.dat")
CE = np.loadtxt(d + "plis_CE.dat")
```

```python
# 24c and 32c gw_fin peak results (each 16 rows = 4 MBGeV x 4 MB; row = iMBGeV*4 + iMB), combined IN
# LOG SPACE. For a positive orders-of-magnitude quantity the natural (multiplicative) uncertainty lives
# in y = ln(x). Per volume: jackknife error propagates to ln as sy = sigma_x / x. Then, independently
# for f_peak (col 3, err 5) and h^2 Omega_peak (col 4, err 6):
#   STEP 1 (jackknife/statistical): inverse-variance mean in log,
#           y_comb = (y24/sy24^2 + y32/sy32^2)/(1/sy24^2 + 1/sy32^2),  s_jk = 1/sqrt(1/sy24^2+1/sy32^2)
#   STEP 2 (systematic): s_syst = |y24 - y32| = |ln(x24/x32)|
#   total: s_tot = sqrt(s_jk^2 + s_syst^2)   (symmetric in ln)
# Central x_comb = exp(y_comb). On the log10 axes below, log10(x_comb) has a SYMMETRIC error bar of
# half-width s_tot/ln(10) -- the multiplicative band expressed directly in log10.
arr24 = np.loadtxt("gw_fin_24c_c2only_c2only_gam3p127_claude.dat")
arr32 = np.loadtxt("gw_fin_v16_32c_c2only_c2only_gam3p127_claude.dat")
_sh24 = np.loadtxt("sigma_hist_24c_v17_claude.dat")  # MBGeV MBrt0 sig_hist_logf sig_hist_logOm (dex)
_sh32 = np.loadtxt("sigma_hist_32c_v17_claude.dat")
_LN10 = np.log(10.0)
_SHCOL = {3: 2, 4: 3}  # value-col -> sigma_hist col (f->2, Omega->3)


def log_combine(cval, cerr):
    x24 = arr24[:, cval]
    e24 = arr24[:, cerr]
    x32 = arr32[:, cval]
    e32 = arr32[:, cerr]
    y24 = np.log(x24)
    y32 = np.log(x32)
    sy24 = e24 / x24
    sy32 = e32 / x32
    sh24 = _sh24[:, _SHCOL[cval]] * _LN10  # histogramming systematic in ln (dex*ln10)
    sh32 = _sh32[:, _SHCOL[cval]] * _LN10
    sy24 = np.sqrt(sy24**2 + sh24**2)  # combined stat (+) hist per volume
    sy32 = np.sqrt(sy32**2 + sh32**2)
    w24 = 1.0 / sy24**2
    w32 = 1.0 / sy32**2
    yc = (w24 * y24 + w32 * y32) / (w24 + w32)
    s_jk = 1.0 / np.sqrt(w24 + w32)
    s_syst = np.abs(y24 - y32)
    s_tot = np.sqrt(s_jk**2 + s_syst**2)
    xc = np.exp(yc)
    return xc, s_jk, s_syst, s_tot


f_c, f_sjk, f_ssyst, f_stot = log_combine(3, 5)
O_c, O_sjk, O_ssyst, O_stot = log_combine(4, 6)

# fT_res_v2 Cell 5 palette: color encodes M_B\sqrt{t_0}, marker encodes M_B [GeV]
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd',
          '#8c564b', '#e377c2', '#7f7f7f', '#bcbd22', '#17becf']
markers = ["o", "d", "x", "v"]
MB_vals = [2.8, 3.2, 3.6, 4.0]
```

```python
# Combined 24c(+)32c sensitivity plot on log10 LINEAR axes -- SAME style as the single-volume N_s plots
# (fT_res_Ns{24,32}_claude): color=M_B\sqrt{t_0}, marker=M_B[GeV]; points+errorbars, no lines, no legend,
# exact Cell-12 arrow annotations, faint grid. Points = the log-space combined estimate; the symmetric
# log error becomes a SYMMETRIC log10 error bar of half-width s_tot/ln(10) (the multiplicative band).
LN10 = np.log(10.0)
lgf = np.log10(f_c)
lgO = np.log10(O_c)
elgf = f_stot / LN10   # combined log10 error on f_peak (jk (+) syst)
elgO = O_stot / LN10   # combined log10 error on h^2 Omega (jk (+) syst)

plt.clf()
fig = plt.figure()
ax = plt.axes()

# barely-visible grid (drawn first, behind everything)
ax.set_axisbelow(True)
ax.grid(True, which='major', color='0.6', alpha=0.30, lw=0.5, zorder=0)
ax.grid(True, which='minor', color='0.8', alpha=0.25, lw=0.4, zorder=0)

# detector threshold bands: PLIS cols are ALREADY log10(f), log10(h^2 Omega) -> plot directly; fill up to 0
ax.fill_between(LISA[::500].T[0], LISA[::500].T[1], 0.0,
                color='C0', edgecolor='b', hatch='X', linewidth=0.0, alpha=0.2)
ax.fill_between(BBO[::400].T[0], BBO[::400].T[1], 0.0,
                color='C1', alpha=0.2)
ax.fill_between(DECIGO[::400].T[0], DECIGO[::400].T[1], 0.0,
                color='C2', edgecolor='g', hatch='|', alpha=0.2)
ax.fill_between(ET[0:160:2].T[0], ET[0:160:2].T[1], 0.0,
                color='C3', edgecolor='C3', hatch='\\', alpha=0.2)
ax.fill_between(CE[:400:5].T[0], CE[:400:5].T[1], 0.0,
                color='C4', edgecolor='C4', hatch='//', alpha=0.2)

# points: color = M_B\sqrt{t_0} (colors[im]), marker = M_B [GeV] (markers[i]); row = i*4 + im
for i in range(4):          # i = M_B [GeV] index (marker)
    for im in range(4):     # im = M_B\sqrt{t_0} index (color)
        r = i * 4 + im
        ax.errorbar(lgf[r], lgO[r], xerr=elgf[r], yerr=elgO[r],
                    ls='none', capsize=3, marker=markers[i], c=colors[im], zorder=3)

ax.set_xlim(-4.0, np.log10(150))
ax.set_ylim(np.log10(2e-20), -12.0)
ax.xaxis.set_major_locator(ticker.MultipleLocator(1))
ax.xaxis.set_minor_locator(ticker.MultipleLocator(0.5))
ax.yaxis.set_major_locator(ticker.MultipleLocator(2))
ax.yaxis.set_minor_locator(ticker.MultipleLocator(1))
plt.xticks(fontsize=22)
plt.yticks(fontsize=22)
ax.set_xlabel("$\\log_{10}(f_{\\rm peak}\\,[{\\rm Hz}])$", fontsize=22)
ax.set_ylabel("$\\log_{10}(h^2\\Omega_{\\rm peak})$", fontsize=22)
ax.set_title("$N_s = 24, 32\\ \\textrm{(combined)}$", fontsize=22)

# --- exact annotations from fT_res_v2 Cell 12 (positions wrapped in log10; rotations unchanged) ---
ax.text(np.log10(0.00022), np.log10(15.6e-19), "$M_B\\sqrt{t_0}=4.0 \\rightarrow$", fontsize=14, color=colors[3])
ax.text(np.log10(0.00038), np.log10(6.4e-19), "$M_B\\sqrt{t_0}=3.6 \\rightarrow$", fontsize=14, color=colors[2])
ax.text(np.log10(2.6), np.log10(4.0e-19), "$\\leftarrow M_B\\sqrt{t_0}=3.2$", fontsize=14, color=colors[1])
ax.text(np.log10(0.0009), np.log10(1.2e-19), "$M_B\\sqrt{t_0}=2.8 \\rightarrow$", fontsize=14, color=colors[0])

# ax.text(np.log10(0.0022), np.log10(2.8e-18), "$M_B=20 \\,{\\rm GeV} \\rightarrow$", fontsize=14, rotation=-50)
# ax.text(np.log10(0.007), np.log10(2.5e-18), "$M_B=70 \\,{\\rm GeV} \\rightarrow$", fontsize=14, rotation=-50)
# ax.text(np.log10(0.017), np.log10(2.5e-18), "$M_B=200 \\,{\\rm GeV} \\rightarrow$", fontsize=14, rotation=-50)
# ax.text(np.log10(0.042), np.log10(2.5e-18), "$M_B=500 \\,{\\rm GeV} \\rightarrow$", fontsize=14, rotation=-50)
ax.text(np.log10(0.0014 / 1.5), np.log10(9.5e-18), "$M_B=20 \\,{\\rm GeV} \\rightarrow$", fontsize=14, rotation=-50)
ax.text(np.log10(0.0042 / 1.5), np.log10(9.5e-18), "$M_B=70 \\,{\\rm GeV} \\rightarrow$", fontsize=14, rotation=-50)
ax.text(np.log10(0.012 / 1.8), np.log10(9.5e-18), "$M_B=200 \\,{\\rm GeV} \\rightarrow$", fontsize=14, rotation=-50)
ax.text(np.log10(0.028 / 1.8), np.log10(9.5e-18), "$M_B=500 \\,{\\rm GeV} \\rightarrow$", fontsize=14, rotation=-50)


ax.text(np.log10(0.00035), np.log10(7.0e-14), "${\\rm LISA}$", fontsize=15, rotation=-50, color='C0')
ax.text(np.log10(0.6), np.log10(1.20e-16), "${\\rm BBO}$", fontsize=15, rotation=0, color='C1')
ax.text(np.log10(0.012), np.log10(125.0e-16), "${\\rm DECIGO}$", fontsize=15, rotation=0, color='C2')
ax.text(np.log10(5.5), np.log10(3.0e-13), "${\\rm ET}$", fontsize=15, rotation=0, color='C3')
ax.text(np.log10(14.0), np.log10(9.0e-14), "${\\rm CE}$", fontsize=15, rotation=0, color='C4')

plt.savefig("sensitivity_24c_32c_comb_logsyst_allMB_c2only_stathist_claude.pdf", format='pdf', bbox_inches='tight')
plt.savefig("sensitivity_24c_32c_comb_logsyst_allMB_c2only_stathist_claude.png", format='png', dpi=150, bbox_inches='tight')
plt.show()
```
